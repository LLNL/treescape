# TreeScape Demo

This notebook demonstrates basic TreeScape functionality with example Caliper data.

```python
import sys
sys.path.append("/app")

import treescape as tr

print("TreeScape imported successfully!")
```

```python
# Load Caliper data
# Mount your data directory when running Docker:
# docker run -v /Users/aschwanden1/datasets/newdemo/test:/app/data ...
cali_file_loc = "/app/data"
processes_for_parallel_read = 4

caliReader = tr.CaliReader(cali_file_loc, processes_for_parallel_read)
print("Caliper data loaded successfully!")
```

```python
# Create TreeScape model
tsm = tr.TreeScapeModel(caliReader)
print(f"TreeScape model created with {len(tsm)} entries")
```
